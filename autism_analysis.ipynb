```python
import pandas as pd
```

```python
df = pd.read_csv("Toddler Autism dataset July 2018.csv")
df.head()
```

```python
# Basic Information from the dataset
df.info()
```

```python
# Shape
print("Rows, Columns:", df.shape)
```

```python
# Checking for missing values
print(df.isnull().sum())
```

```python
# Quick Stats
df.describe(include='all')
```

```python
# Data Cleaning
# Droping duplicates if any
df = df.drop_duplicates()
```

```python
# Exploratory Data Analysis (EDA)
# Demographics
import matplotlib.pyplot as plt
import seaborn as sns

# Gender Distribution
sns.countplot(x='Sex', data=df, palette='viridis')
plt.title("Toddlers who were screened")
plt.xlabel("Sex")
plt.ylabel("Count")
plt.show()
```

```python
# Gender vs Autism Diagnosis
sns.countplot(x='Sex', hue="Class/ASD Traits ", data=df, palette="Set2")
plt.title("Gender vs Autism Diagnosis")
plt.xlabel("Sex")
plt.ylabel("Count")
plt.show()
```

```python
# Age Distribution
sns.histplot(df['Age_Mons'], bins=20, kde=True, color="green")
plt.title("Age DIstribution (Months)")
plt.xlabel("Age in Months")
plt.ylabel("Frequency")
plt.show()
```

```python
# Ethinicity Distribution
plt.figure(figsize=(10, 5))
df['Ethnicity']. value_counts().plot(kind='bar', color="orange")
plt.title("Ethnicity Distribution")
plt.xlabel("Ethnicity")
plt.ylabel("Count")
plt.show()
```

```python
# Ethnicity vs Autism Diagnosis
plt.figure(figsize=(12, 6))
sns.countplot(x="Ethnicity", hue="Class/ASD Traits ", data=df, palette="coolwarm")
plt.xticks(rotation=45)
plt.title("Autism Diagnosis by Ethnicity")
plt.show()
```

```python
# Family History & Autism Diagnosis
sns.countplot(x='Family_mem_with_ASD', hue='Class/ASD Traits ', data=df, palette="Set1")
plt.title("Family HIstory of ASD vs Autism DIagnosis")
plt.xlabel("Family History of ASD")
plt.ylabel("Count")
plt.show()
```

```python
# Jaundice & Autism Diagnosis
sns.countplot(x='Jaundice', hue='Class/ASD Traits ', data=df, palette="Set3")
plt.title("Jaundice at Birth vs Autism Diagnosis")
plt.xlabel("Jaundice at Birth")
plt.ylabel("Count")
plt.show()
```

```python
# Screening Score
sns.histplot(df, x="Qchat-10-Score", hue="Class/ASD Traits ", kde=True, bins=15, palette="husl")
plt.title("Q-CHAT 10 Score Distribution by Autism Diagnosis")
plt.xlabel("Q-CHAT 10 Score")
plt.ylabel("Count")
plt.show()
```
